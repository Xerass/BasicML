```python

import re
import nltk
import os

# Ensure NLTK data directory exists
nltk_data_path = os.path.expanduser('~/nltk_data')
os.makedirs(nltk_data_path, exist_ok=True)
if nltk_data_path not in nltk.data.path:
    nltk.data.path.append(nltk_data_path)

# Required NLTK resources and their paths
required_packages = {
    'punkt': 'tokenizers/punkt',
    'stopwords': 'corpora/stopwords',
    'wordnet': 'corpora/wordnet',
    'averaged_perceptron_tagger': 'taggers/averaged_perceptron_tagger'
}

# Download only what is missing
for pkg, location in required_packages.items():
    try:
        nltk.data.find(location)
        print(f"{pkg} already installed.")
    except LookupError:
        print(f"Downloading {pkg}...")
        nltk.download(pkg, download_dir=nltk_data_path, quiet=True)

# Imports after data is confirmed
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
import spacy
```

-stopwords are for common words like a, the, is as they add no semantic meaning to text
-lemmatizer reduces words to dict form for easier meaning retrieval 
-stemmer cuts off word endings

```python
#load a spacy model
nlp = spacy.load("en_core_web_sm")

#create a sample dataset
texts = [
    "Natural Language Processing (NLP) is AMAZING!!!",
    "Python proviides powerful tools for text cleaning and tokenization.",
    "Students should complete their pre-processing tasks on time.",
    "This text is used to test the ability of the model to pre-process data."
]

print("Original Text before Pre-Processing:")
for i in texts:
    print("-", i)
```

```python
#instantiate the pre-processing tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

#removes all special chars and numbers
def clean_text(text):
    text = re.sub(r"[^a-zA-Z\s]", "",text)
    return text

def preprocess_text(text):
    print("\n==================================================")
    
    #perform data cleaning
    cleaned = clean_text(text)
    print("Cleaned Text:",cleaned)

    #token
    tokens = word_tokenize(cleaned)
    print("Tokens:", tokens)

    #lowercase
    lowercased = [t.lower() for t in tokens]
    print("Lowercased Tokens:", lowercased)

    #stop words removal
    no_stop = [t for t in lowercased if t not in stop_words]
    print("No Stop Words:", no_stop)

    #lemmatization
    lemmas = [lemmatizer.lemmatize(t) for t in no_stop]
    print("Lemmatized Tokens:", lemmas)

    #stemming
    stems = [stemmer.stem(t) for t in no_stop]
    print("Stemmed:", stems)

    #POS tagging
    doc = nlp(' '.join(tokens))
    pos_tags = [(token.text, token.pos) for token in doc]
    print("POS tags:", pos_tags)
```

```python
#apply it to text
for t in texts:
    preprocess_text(t)
```
